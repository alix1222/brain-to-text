# Top-1

```python
from datasets import load_dataset, Dataset

# 1. Load dataset
dataset_1 = load_dataset("json", data_files="val_candidates_seed1.json", split="train")
dataset_2 = load_dataset("json", data_files="val_candidates_seed6.json", split="train")
dataset_3 = load_dataset("json", data_files="val_candidates_seed9.json", split="train")
dataset_4 = load_dataset("json", data_files="val_candidates_seed11.json", split="train")
dataset_5 = load_dataset("json", data_files="val_candidates_seed15.json", split="train")
dataset_6 = load_dataset("json", data_files="val_candidates_seed17.json", split="train")
dataset_7 = load_dataset("json", data_files="val_candidates_seed20.json", split="train")
dataset_8 = load_dataset("json", data_files="val_candidates_seed21.json", split="train")
dataset_9 = load_dataset("json", data_files="val_candidates_seed26.json", split="train")
dataset_10 = load_dataset("json", data_files="val_candidates_seed28.json", split="train")
```

```python
print(f"Dataset 1 samples: {len(dataset_1)}")
print(f"Dataset 2 samples: {len(dataset_2)}")
print(f"Dataset 3 samples: {len(dataset_3)}")
print(f"Dataset 4 samples: {len(dataset_4)}")
print(f"Dataset 5 samples: {len(dataset_5)}")
print(f"Dataset 6 samples: {len(dataset_6)}")
print(f"Dataset 7 samples: {len(dataset_7)}")
print(f"Dataset 8 samples: {len(dataset_8)}")
print(f"Dataset 9 samples: {len(dataset_9)}")
print(f"Dataset 10 samples: {len(dataset_10)}")

# 2. Merge top-1 candidates + ground truth
all_candidates = []
all_ground_truths = []

for idx in range(len(dataset_1)):
    # Get top-1 candidate from each dataset
    top1_seed1 = dataset_1[idx]["candidates"][0]
    top1_seed2 = dataset_2[idx]["candidates"][0]
    top1_seed3 = dataset_3[idx]["candidates"][0]
    top1_seed4 = dataset_4[idx]["candidates"][0]
    top1_seed5 = dataset_5[idx]["candidates"][0]
    top1_seed6 = dataset_6[idx]["candidates"][0]
    top1_seed7 = dataset_7[idx]["candidates"][0]
    top1_seed8 = dataset_8[idx]["candidates"][0]
    top1_seed9 = dataset_9[idx]["candidates"][0]
    top1_seed10 = dataset_10[idx]["candidates"][0]
    
    # Get ground truth
    ground_truth = dataset_1[idx]["ground_truth"]
    
    # Combine into candidates list (top-1 from each + ground truth)
    # Remove duplicates while preserving order
    candidates = []
    # seen = set()
    
    for cand in [top1_seed1, top1_seed2, top1_seed3, top1_seed4, top1_seed5, top1_seed6, top1_seed7, top1_seed8, top1_seed9, top1_seed10]:
        # if cand not in seen:
        candidates.append(cand)
            # seen.add(cand)
    
    all_candidates.append(candidates)
    all_ground_truths.append(ground_truth)

# 3. Create merged dataset
dataset_all = Dataset.from_dict({
    "candidates": all_candidates,
    "ground_truth": all_ground_truths
})

print(f"\n✅ Merged dataset created!")
print(f"Total samples: {len(dataset_all)}")
print(f"\nExample merged sample:")
print(f"Candidates: {dataset_all[0]['candidates']}")
print(f"Ground truth: {dataset_all[0]['ground_truth']}")
print(f"Number of unique candidates: {len(dataset_all[0]['candidates'])}")
```

```python
# Save the merged dataset to JSON
dataset_all.to_json("val_candidates_seed1_6_9_11_15_17_20_21_26_28_top1.json")

print(f"✅ Dataset saved to 'val_candidates_seed1_6_9_11_15_17_20_21_26_28_top1.json'")
print(f"Total samples saved: {len(dataset_all)}")
```
